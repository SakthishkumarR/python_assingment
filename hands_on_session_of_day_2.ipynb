````python
# ## Section 1: Control Structures (If-Else Statements)

# ### Problem 1.1: Grade Calculator (Easy)

# Write a program that takes a student's marks (0-100) as input and prints their grade according to the following criteria:
# - 90-100: Grade A
# - 80-89: Grade B
# - 70-79: Grade C
# - 60-69: Grade D
# - Below 60: Grade F

# # Also, validate that the marks are within the valid range (0-100).

numbers = int(input("Enter your score here : "))

if numbers >= 90 and numbers <= 100:
  print(f"the grade for the marks {numbers} is 'A' ")

elif numbers >= 80 and numbers <= 89:
  print(f"the grade for the marks {numbers} is 'B' ")

elif numbers >= 70 and numbers <= 79:
  print(f"the grade for the marks {numbers} is 'C' ")

elif numbers >= 60 and numbers <= 69:
  print(f"the grade for the marks {numbers} is 'D' ")

elif numbers < 60 and numbers >= 0:
  print(f"the grade for the marks {numbers} is 'F' ")

else:
  print("invalid input")

````

````python

# Write a program to check if a given year is a leap year or not.

# Rules:
# - A year is a leap year if it is divisible by 4
# - However, if it is divisible by 100, it is NOT a leap year
# - Exception: If it is also divisible by 400, then it IS a leap year

# Examples:
# - 2000: Leap year (divisible by 400)
# - 1900: Not a leap year (divisible by 100 but not 400)
# - 2024: Leap year (divisible by 4 but not 100)

# **Hint:** Check conditions in order: first divisible by 400, then by 100, then by 4. Use modulo operator (%) to check divisibility.



year = int(input("enter a year here : "))

if year % 400 == 0:
  print(f"{year} is a leap year ")

elif year % 100 == 0:
  print(f"{year} is not a leap year")

elif year % 4 == 0:
  print(f"{year} is a leap year")

else:
  print(f"{year} is not a leap year")
````

````python
# ### Problem 1.3: Triangle Type Identifier (Medium)

# Write a program that takes three sides of a triangle as input and determines:
# 1. Whether the three sides can form a valid triangle (sum of any two sides must be greater than the third side)
# 2. If valid, classify the triangle as:
#    - Equilateral (all three sides equal)
#    - Isosceles (exactly two sides equal)
#    - Scalene (no sides equal)


a = int(input("enter your first side here : "))
b = int(input("enter your second side here : "))
c = int(input("enter your third side here : "))

if a + b > c and b + c > a and c + a > b:
  if a == b == c:
    print("the triangle is equilateral")
  elif a == b or b == c or c == a:
    print("the triangle is isosceles")
  else:
    print("the triangle is scalene")
else:
  print("the triangle is not valid")
````

````python
# ### Problem 2.2: Factorial Calculator (Medium)

# Write a program to calculate the factorial of a number using a for loop.

# Factorial of n (n!) = n × (n-1) × (n-2) × ... × 1

# Example: 5! = 5 × 4 × 3 × 2 × 1 = 120

# **Hint:** Initialize result = 1. Use for loop with range(1, n+1) and multiply result by each number.


num = int(input("enter a number you want factorial of : "))

result = 1

for i in range(1,num+1):
  result = result * i

print(f"the factorial of {num} is {result} ")
````

````python

# Write a program to print the first n terms of the Fibonacci series.

# The Fibonacci series: 0, 1, 1, 2, 3, 5, 8, 13, 21, ...

# Each number is the sum of the two preceding ones.

# **Hint:** Initialize a=0, b=1. Print a, then in loop: calculate next=a+b, print next, then update a=b and b=next.



n = int(input("Enter the number of terms: "))

a = 0
b = 1

for i in range(n):
    print(a, end=" ")
    next = a + b
    a = b
    b = next
````

````python

# ### Problem 3.1: Reverse a Number (Medium)

# Write a program that reverses the digits of a given number using a while loop.

# Example: Input: 12345, Output: 54321

# **Hint:** Initialize reverse=0. While n>0: get last digit (n%10), add to reverse (reverse=reverse*10+digit), remove last digit (n=n//10).

numbers = int(input("enter the number you want to reverse : "))

temp = numbers

reverse = 0

while temp > 0:
    digit = temp % 10
    reverse = reverse * 10 + digit
    temp = temp // 10

print(f"the reverse number of {numbers} is {reverse}")
````

````python
# ### Problem 3.2: Sum of Digits (Easy)

# Write a program that calculates the sum of all digits in a given number.

# Example: Input: 12345, Output: 15 (1+2+3+4+5)

# **Hint:** While n>0: add last digit to sum (sum += n%10), then remove it (n = n//10).


n = 12345
total = 0

while n > 0:
    digit = n % 10
    total = total + digit
    n = n // 10

print(total)


````

````python
# ### Problem 3.3: Number Guessing Game (Medium)

# Write a number guessing game where:
# 1. The program generates a random number between 1 and 100
# 2. The user has to guess the number
# 3. After each guess, the program tells the user if their guess is too high or too low
# 4. The game continues until the user guesses correctly
# 5. Display the number of attempts taken

# **Hint:** Use `import random` and `random.randint(1, 100)` to generate number. Use while loop with condition guess != target. Count attempts.



import random

target = random.randint(1, 100)
guess = 0
attempts = 0

while guess != target:
    guess = int(input("Guess a number between 1 and 100: "))
    attempts = attempts + 1

    if guess > target:
        print("Too high! Try again.")
    elif guess < target:
        print("Too low! Try again.")
    else:
        print("Correct! You guessed the number.")
        print("Attempts:", attempts)
````

````python
# ### Problem 3.4: Palindrome Checker (Medium)

# Write a program to check if a given number is a palindrome or not.

# A palindrome number reads the same forwards and backwards.

# Examples: 121, 12321, 1331 are palindromes

# **Hint:** Store original number, reverse it (like Problem 3.1), then compare original with reversed number.


n = input("enter the number to check if it is a palindrome : ")


reverse = n[::-1]

if n == reverse:
    print(f"the number you entered as {n} is a  Palindrome number")
else:
    print("the number you entered is Not a palindrome number")
````
