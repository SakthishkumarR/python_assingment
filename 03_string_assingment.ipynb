```python
##que 41

morse = {
    "h": "....",
    "e": ".",
    "l": ".-..",
    "o": "---",
    "w": ".--",
    "r": ".-.",
    "d": "-.."
}

text = "hello world"

result = ""

for letter in text:
    if letter == " ":
        result += " / "
    else:
        result += morse[letter] + " "

print(result)
```

```python
##que 42

text1 = "programming"
text2 = "grammar"

longest = ""

for i in range(len(text1)):
    for j in range(len(text2)):
        temp = ""

        while (i + len(temp) < len(text1) and
               j + len(temp) < len(text2) and
               text1[i + len(temp)] == text2[j + len(temp)]):
            temp += text1[i + len(temp)]

        if len(temp) > len(longest):
            longest = temp

print(longest)
```
