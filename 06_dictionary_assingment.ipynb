```python
## Question 49: Create a dictionary with nested sets


my_dict = {
    'groups': [{1, 2, 3}, {4, 5, 6}],
    'categories': [{'a', 'b'}, {'c', 'd'}]
}
print(my_dict)
```

```python
## Question 50: Find union of all nested sets


groups_union = my_dict['groups'][0] | my_dict['groups'][1]
categories_union = my_dict['categories'][0] | my_dict['categories'][1]

print(groups_union)
print(categories_union)
```
