```python
# 1. Right-angled triangle — stars


n = 5
for i in range(1,n+1):
  print("*" * i )
```

```python
# 2. Inverted right-angled triangle — stars

n = 5

for i in range(n , 0 , -1):
  print("*" * i)
```

```python
# 3. Solid square of stars

n = 3


for i in range(n):
  print("*" * n)
```

```python
# 4. Hollow square

n = 5
for i in range(n):
    if i == 0 or i == n - 1:
        print("*" * n)
    else:
        print("*" + " " * (n - 2) + "*")
```

```python
# 5. Number triangle (1, 12, 123, …)

n = 5

for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end="")
    print()
```

```python

# 6. Floyd’s triangle (continuous numbers)

n = 7
for i in range(1,n+1):
  for j in range(1,i+1):
    print(j,end="")
  print()
```

```python
n = 4
num = 1
for i in range(1, n + 1):
    for j in range(i):
        print(num, end=" ")
        num = num + 1
    print()
```

```python
n = 4
output  = 1

for i in range(1,n+1):
  for j in range(i):
    print(output, end = "")
    output = output +1
  print()
```

```python
# 7. Centered star pyramid


n = 4
for i in range(1, n + 1):
    spaces = " " * (n - i)
    stars = "*" * (2 * i - 1)
    print(spaces + stars)
```
